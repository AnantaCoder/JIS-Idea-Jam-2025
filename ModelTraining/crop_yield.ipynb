```python
import numpy as np 
import pandas as pd
import seaborn as sns 
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('datasets/crop_yield.csv')
dataset['Season'] = dataset['Season'].str.strip().str.title()
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:, -1].values
print(dataset.head())
```

```python
print(dataset.isnull().sum()) # no missing values 
```

we have few categorical and non categorical data s 
-> categorical => crop , season , state 
-> non categorical => yield , price , area, production , annual rainfall , fertilizer , pesticide and yield 

```python
categorical_features = ['Crop', 'Season', 'State']
numerical_features = ['Crop_Year', 'Area', 'Annual_Rainfall', 'Fertilizer', 'Pesticide']
target_variable = 'Yield'
```

```python
X = dataset[categorical_features + numerical_features]  
y = dataset[target_variable]  
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler


preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('num', StandardScaler(), numerical_features)
    ])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor as rfrg
model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('regressor', rfrg(
        n_estimators=200,
        max_depth=100,
        random_state=42
    ))
])
# incrsed the depth from 20 to 40
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Make predictions
y_pred = model.predict(X_test)

# Evaluate regression performance
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"R² Score: {r2:.2f}") 
```

```python
import joblib
joblib.dump(model,'Modelss/CropYieldPrediction.pkl')
```
